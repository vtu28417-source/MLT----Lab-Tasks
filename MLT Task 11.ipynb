```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    classification_report, confusion_matrix
)

# Step 1: Dataset
documents = [
    "Python programming and software development",
    "Java programming and object oriented concepts",
    "Software engineering and application development",
    "Python code and machine learning algorithms",
    "Computer programming using Java and Python",
    "Software testing and debugging applications",
    "Neural networks and artificial intelligence",
    "Machine learning models and deep learning",
    "Artificial intelligence and neural networks",
    "Deep learning algorithms for image recognition",
    "Cricket match and sports tournament",
    "Football players and international matches",
    "Cricket team won the championship",
    "Football tournament and sports competition",
    "Sports players practice for cricket matches",
    "Cricket world cup and team performance",
    "Football league and team championship",
    "Athletes participate in sports competitions",
    "Cricket players and international tournament",
    "Football match and sports training",
    "Hospital doctors and patient treatment",
    "Medical diagnosis and healthcare services",
    "Doctors provide treatment to patients",
    "Hospital management and medical services",
    "Healthcare professionals and patient care",
    "Medical research and disease treatment",
    "Patient diagnosis in healthcare centers",
    "Doctors and nurses provide hospital care",
    "Healthcare system and medical facilities",
    "Disease prevention and patient treatment"
]

labels = (
    ["Technology"] * 10 +
    ["Sports"] * 10 +
    ["Healthcare"] * 10
)

# Step 2: Convert text into numerical features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(documents)
y = np.array(labels)

# Step 3: Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30,
    random_state=42, stratify=y
)

# Step 4: Train RBF classifier
model = SVC(kernel="rbf", C=1.0, gamma="scale")
model.fit(X_train, y_train)

# Step 5: Prediction
y_pred = model.predict(X_test)

# Step 6: Calculate metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(
    y_test, y_pred, average="macro", zero_division=0
)
recall = recall_score(
    y_test, y_pred, average="macro", zero_division=0
)
classification_rate = np.mean(y_test == y_pred)

# Step 7: Display results
print("RBF CLASSIFIER - DOCUMENT CLASSIFICATION")
print("-----------------------------------------")
print("Total documents:", len(documents))
print("Training documents:", len(y_train))
print("Testing documents:", len(y_test))

print("\nActual Labels:", list(y_test))
print("Predicted Labels:", list(y_pred))

print("\nClassification Rate: {:.2f}%".format(
    classification_rate * 100))
print("Accuracy: {:.2f}%".format(accuracy * 100))
print("Precision: {:.2f}%".format(precision * 100))
print("Recall: {:.2f}%".format(recall * 100))

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test, y_pred,
    labels=["Technology", "Sports", "Healthcare"]
))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
